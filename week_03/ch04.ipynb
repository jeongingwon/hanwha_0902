```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(
    temperature=0.1,
    model_name="gpt-4o-mini"
)

question = "대한민국의 수도는 어디인가요?"

response = llm.invoke(question)
```

```python
print(f"[답변]:{response}")
```

```python
response.response_metadata["token_usage"]
```

```python
response.response_metadata["token_usage"]["total_tokens"]
```

### logprob 로그 프로버블리티 활성화
- 모델의 답변 확신도(Cofidence level):0에 가까울수록 정답에 가까운 확신을 가진다

```python
#obj 생성
llm_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_completion_tokens=2048,
    model="gpt-4o-mini",
).bind(logprobs=True)
```

```python
# 질의내용
question = "인도의 수도는 어디인가요?"

# 질의
response = llm_with_logprob.invoke(question)
```

```python
response.response_metadata
```

```python
response.response_metadata["logprobs"]
```

```python
#답변을. 한토큰씩 받아와서 화면에 실시간으로 뿌려줄때

answer = llm.stream("미국의 아름다운 관광지 10곳과 주소를 알려주세요!")
```

```python
#flush=False: 결과가 버퍼에 저장되었다가. 한번에 출력
#flush=True : 결과가 생성되는대로 바로바로 출력
for token in answer:
    print(token.content, end="", flush=True)
```
